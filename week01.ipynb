#### এই সপ্তাহে Theory ও Math-এর সম্পর্ক

পদার্থবিজ্ঞানের বইয়ে এক পরমাণুর আকার কেউ ০.০০০০০০০০০১ মিটার লেখে না, লেখে 10⁻¹⁰ মিটার। এত শূন্য গুনে গুনে লিখলে ভুল হওয়ার ভয় থাকে, তাই খুব ছোট মাপ আমরা সহজ রূপে লিখি।

এই সপ্তাহের Theory-তে তুমি পড়বে AI, LLM ও Next Token Prediction। LLM যখন একটা বাক্য লেখে, তখন প্রতিটা ধাপে সে পরের token-এর জন্য একটা সম্ভাবনা (probability) হিসাব করে। সেই সংখ্যাটা সাধারণত খুব ছোট হয়, যেমন ০.০০০২। পুরো বাক্যের সম্ভাবনা পেতে এমন অনেক ছোট সংখ্যা একটার সাথে আরেকটা গুণ করতে হয়, আর গুণ করতে করতে ফলাফল ছোট হতে হতে প্রায় শূন্য হয়ে যায়।

এখানেই Math কাজে আসে। Logarithm এই ছোট সংখ্যাগুলোকে সামলানো সহজ সংখ্যায় বদলে দেয়, ঠিক যেমন পরমাণুর আকার 10⁻¹⁰ লিখলে পড়তে ও হিসাব করতে সুবিধা হয়। তাই Algebra, Function, Exponent ও Logarithm আলাদা কোনো বিষয় নয়, এগুলো এই সপ্তাহের Theory বোঝার সরাসরি হাতিয়ার।

> **সারকথা:** Probability হলো LLM-এর ভাষা, আর Logarithm সেই ছোট সংখ্যাগুলো পড়ার সহজ উপায়।

#### Part 1: AI Theory

#### Artificial Intelligence (AI) কী

#### আগে একটা গল্প
তোমার email-এ রোজ অনেক চিঠি আসে। কিছু আসল, কিছু বিজ্ঞাপনের জঞ্জাল, যাকে বলে **spam** (অবাঞ্ছিত, বিরক্তিকর email)। এই জঞ্জাল আলাদা করবে কে?

প্রথম উপায়: একজন programmer বসে নিয়ম লেখে। "চিঠিতে 'লটারি জিতেছেন' থাকলে spam।" কিন্তু প্রতারকরা রোজ নতুন কথা লেখে। নিয়মের তালিকা কখনো শেষ হয় না।

দ্বিতীয় উপায়: computer-কে হাজার হাজার আসল আর spam চিঠি দেখানো হয়। সে নিজেই বুঝে নেয়, কোন ধরনের চিঠি spam। নতুন ধরনের spam এলেও সে ধরে ফেলে।

#### মূল ধারণা
**Software** মানে computer-এর program, অর্থাৎ computer-কে দেওয়া নির্দেশের তালিকা। **Artificial Intelligence (AI)**, বাংলায় কৃত্রিম বুদ্ধিমত্তা, হলো এমন software, যা মানুষের বুদ্ধি লাগে এমন কাজ করতে পারে। যেমন ছবি চেনা, ভাষা বোঝা, কথার উত্তর দেওয়া।

সাধারণ program আর AI-র তফাত একটাই:

> সাধারণ program → নিয়ম লিখে দেয় মানুষ  
> AI → **data** (তথ্য) দেখে নিয়ম শেখে নিজে

#### মনে রাখার এক লাইন
> সাধারণ program-এ নিয়ম দেয় মানুষ। AI নিজে data দেখে নিয়ম শেখে।

#### Machine Learning vs Deep Learning

#### আগে একটা গল্প
একটা ছোট শিশুকে কীভাবে বিড়াল চেনানো হয়? কেউ তাকে বলে না, "যার চারটা পা, লম্বা গোঁফ আর খাড়া কান, সেটাই বিড়াল।" বরং সে রাস্তায়, বাড়িতে, ছবির বইয়ে অনেক বিড়াল দেখে। কিছুদিন পর সে নিজেই বুঝে যায়, "এটা বিড়াল, এটা কুকুর।"

**Machine Learning (ML)** ঠিক এভাবেই কাজ করে। আমরা computer-কে নিয়ম লিখে দিই না। তাকে হাজার হাজার **উদাহরণ** দেখাই, আর সে নিজে নিজে **pattern** (বারবার ফিরে আসা মিল) খুঁজে নেয়।

#### তাহলে Deep Learning কী?
ML-এ computer-কে বিড়াল চেনাতে হলে মানুষকে আগে বলে দিতে হয়, "কান দেখো, গোঁফ দেখো, লেজ দেখো।" এই বেছে দেওয়ার কাজটা মানুষের। **Deep Learning (DL)** এই কাজটাও নিজে করে। তুমি শুধু কাঁচা ছবি দাও। ভেতরে অনেক **স্তর (layer)** সাজানো একটা নেটওয়ার্ক থাকে, যাকে বলে **neural network**। প্রতিটা স্তর একটু একটু করে বোঝে:

> প্রথম স্তর → রেখা ও কিনারা  
> পরের স্তর → বাঁকা আকার  
> তার পরের স্তর → কান, চোখ, গোঁফ  
> শেষ স্তর → "এটা বিড়াল"

স্তর অনেক বলেই এর নাম **Deep** (গভীর) Learning।

#### তিনটা শব্দের সম্পর্ক
তিনটা বৃত্ত, একটার ভেতরে আরেকটা: **AI ⊃ ML ⊃ DL**। AI সবচেয়ে বড়, ML তার ভেতরে, DL সবচেয়ে ভেতরে। আজকের ChatGPT বা Claude এই Deep Learning-এরই ফসল।

#### মনে রাখার এক লাইন
> ML উদাহরণ দেখে শেখে। DL সেই শেখাটা অনেক স্তরে, নিজে নিজে করে।

#### Generative AI ও Large Language Model (LLM)

#### আগে একটা গল্প
কল্পনা করো, এক ছাত্র একটা বিশাল লাইব্রেরির প্রায় সব বই পড়ে ফেলেছে। উপন্যাস, বিজ্ঞান, ইতিহাস, রান্নার বই, সব। এখন তুমি তাকে যেকোনো বিষয়ে লিখতে বললে সে লিখে দিতে পারে। সে বই থেকে হুবহু টুকে দেয় না। যা পড়েছে, তার ভাষা আর জ্ঞান মিলিয়ে নতুন করে লেখে।

#### Generative AI কী?
**Generative AI** মানে তৈরিকারী AI। সে নতুন কিছু বানায়: লেখা, ছবি, গান বা code। আগের অনেক AI শুধু চিনত বা আলাদা করত, যেমন spam চেনা। Generative AI বানায়।

#### LLM কী?
**Large Language Model (LLM)** মানে বিশাল ভাষা-মডেল। **Model** হলো শেখা শেষে তৈরি হওয়া AI-র মূল অংশ। LLM হলো এমন model, যে বিশাল পরিমাণ লেখা পড়ে ভাষা শিখেছে এবং এখন লেখা তৈরি করতে পারে। গল্পের সেই ছাত্রই LLM, আর লাইব্রেরির বইগুলো তার শেখার data।

**ChatGPT** আর **Claude** এই LLM-এর উদাহরণ।

#### দুটো শব্দের সম্পর্ক
Generative AI একটা বড় পরিবার। তার ভেতরে যে সদস্য লেখা বানায়, তার নাম LLM। ছবি বানানোর AI আলাদা সদস্য।

> Generative AI → লেখা, ছবি, গান বানায়  
> LLM → তার ভেতরে, শুধু লেখা বানায়

#### মনে রাখার এক লাইন
> Generative AI নতুন কিছু বানায়। LLM হলো বিশাল লেখা পড়ে শেখা model, যে লেখা বানায়।

#### Training vs Inference

#### আগে একটা গল্প
রিমা তিন মাস ধরে পরীক্ষার জন্য পড়ছে। বই পড়ছে, অঙ্ক কষছে, ভুল করছে, ভুল শুধরাচ্ছে। এই তিন মাসে তার মাথার ভেতরের জ্ঞান বদলাচ্ছে।

পরীক্ষার দিন সে হলে বসে প্রশ্ন দেখে আর উত্তর লেখে। এখন সে নতুন কিছু শিখছে না। যা শিখেছে, তাই কাজে লাগাচ্ছে।

#### দুটো পর্যায়
**Training** (প্রশিক্ষণ) হলো পড়াশোনার সময়। Model অনেক data দেখে, ভুল করে, আর নিজের ভেতরের সংখ্যাগুলো একটু একটু করে বদলে নেয়। এতে অনেক সময়, অনেক computer আর অনেক টাকা লাগে।

**Inference** (শেখা বিদ্যা দিয়ে উত্তর দেওয়া) হলো পরীক্ষার হলের সময়। Model প্রশ্ন পায়, উত্তর দেয়। ভেতরের কোনো সংখ্যা বদলায় না।

> Training → শেখো (ভেতরের সংখ্যা বদলায়)  
> Inference → উত্তর দাও (কিছুই বদলায় না)

#### তুমি কোনটা করছ?
**API** মানে এক program-এর সাথে আরেক program-এর কথা বলার দরজা। API দিয়ে যখন LLM-কে প্রশ্ন পাঠাও, তখন তুমি **inference** করছ। তুমি model-কে শেখাচ্ছ না। তার শেখা বিদ্যা কাজে লাগাচ্ছ।

#### মনে রাখার এক লাইন
> Training হলো শেখা, Inference হলো শেখা বিদ্যা ব্যবহার। API-কে প্রশ্ন পাঠানো মানে inference।

#### Pre-trained Models ও Model Parameters

#### আগে একটা গল্প
তুমি মাংসের ঝোল রাঁধবে। এক উপায়: ধনে, জিরা, হলুদ, মরিচ আলাদা করে কিনে, শুকিয়ে, বেটে নিজে মশলা বানানো। অনেক সময় লাগে। আরেক উপায়: দোকান থেকে তৈরি মশলা কিনে এনে সরাসরি রান্নায় দেওয়া।

AI-র বেলাতেও তাই। Model বানানো (training) খুব খরচের কাজ। তাই বেশিরভাগ মানুষ কারও বানানো model-ই ব্যবহার করে।

#### Pre-trained model
**Pre-trained model** মানে আগে থেকে শেখানো model। কোনো বড় প্রতিষ্ঠান বিশাল data আর অনেক computer দিয়ে তাকে শিখিয়ে রেখেছে। আমরা শুধু ব্যবহার করি, যেমন তৈরি মশলা দিয়ে রান্না করি।

#### Parameters
Model তার শেখা জ্ঞান রাখে কোথায়? **Parameters**-এ। Parameter হলো অনেকগুলো সংখ্যা। মস্তিষ্কে স্নায়ুকোষের সংযোগ যেমন জ্ঞান ধরে রাখে, model-এ এই সংখ্যাগুলো তেমনি জ্ঞান ধরে রাখে। Training-এর সময় এই সংখ্যাগুলোই বদলায়।

বড় model-এ এমন সংখ্যা থাকে কয়েক **বিলিয়ন** থেকে শত শত বিলিয়ন। বিলিয়ন মানে শত কোটি।

> Training → parameter-এর সংখ্যা বদলে বদলে জ্ঞান জমে  
> Pre-trained model → সেই জ্ঞান জমা হয়ে আছে, আমরা ব্যবহার করি

#### মনে রাখার এক লাইন
> Pre-trained মানে আগেই শেখানো। Parameters হলো সেই শেখা জ্ঞান ধরে রাখা সংখ্যা।

#### Next Token Prediction

#### আগে একটা গল্প
মোবাইলে message লেখার সময় তুমি "আমি" লিখলে keyboard-এর ওপরে কয়েকটা শব্দ ভেসে ওঠে। তুমি যেটা চাও, ছুঁয়ে দাও। Keyboard তোমার কথা বোঝে না। সে শুধু আন্দাজ করে, এরপর কোন শব্দ আসার সম্ভাবনা সবচেয়ে বেশি।

#### LLM-ও একই কাজ করে
LLM আসলে এই কাজটাই করে, শুধু অনেক বেশি স্মার্টভাবে। সে একবারে একটা করে পরের টুকরো অনুমান করে। এই টুকরোর নাম **token** (লেখার ছোট টুকরো, যেমন একটা শব্দ বা শব্দের অংশ)।

প্রতিটা ধাপে সে প্রতিটা সম্ভাব্য পরের token-এর **probability** (সম্ভাবনা, অর্থাৎ কতটা সম্ভব তার মাপ) হিসাব করে। তারপর সবচেয়ে সম্ভাব্যটা বেছে নেয়। পরের সপ্তাহে দেখব, বাছাইয়ে একটু এলোমেলোভাবও আনা যায়।

> "আমি ভাত" → খাই (70%), খাচ্ছি (20%), রাঁধি (7%) ...  
> বেছে নিল → "খাই"  
> "আমি ভাত খাই" → আবার পরের token অনুমান ...

এভাবে একটা একটা করে token জুড়ে পুরো উত্তর তৈরি হয়।

**Demo:** নিচের code-এ "আমি ভাত"-এর পরে কোন শব্দের কত সম্ভাবনা, তা একটা ছোট তালিকায় রেখে সবচেয়ে সম্ভাব্যটা বেছে দেখানো হয়েছে।

#### মনে রাখার এক লাইন
> LLM একবারে একটা করে পরের token অনুমান করে, ঠিক keyboard-এর suggestion-এর মতো।

#### 1. Order of Operations

**মানে:** আগে Bracket, তারপর Power, তারপর Multiply/Divide, শেষে Add/Subtract।

**Example:** `2 + 3 × 4 = 14` (আগে গুণ)। কিন্তু `(2 + 3) × 4 = 20` (আগে bracket)।

**AI-তে কেন লাগে:** `w * x + b` ভুল ক্রমে করলে model ভুল উত্তর দেয়।

```python
print(2 + 3 * 4)       # 14  (আগে গুণ)
print((2 + 3) * 4)     # 20  (আগে bracket)
print(2 + 3 ** 2)      # 11  (আগে power)
assert 2 + 3 * 4 == 14 and (2 + 3) * 4 == 20 and 2 + 3 ** 2 == 11
```

<a id="sec2"></a>
#### 2. Fractions

**মানে:** Fraction হলো ভাগের সংখ্যা। উপরেরটা numerator, নিচেরটা denominator। যোগে denominator সমান করতে হয়, ভাগে উল্টে গুণ করতে হয়।

**Example:** `1/2 + 1/3 = 5/6`, `1/2 × 1/3 = 1/6`, `1/2 ÷ 1/3 = 3/2`, `6/8 = 3/4` (simplest form)।

**AI-তে কেন লাগে:** decimal-এ ছোট ভুল থাকে (`0.1 + 0.2 ≠ 0.3`), exact fraction-এ থাকে না। ছোট ভুল জমে বড় হয়।

```python
a, b = Fraction(1, 2), Fraction(1, 3)
print("add      :", a + b)            # 5/6
print("multiply :", a * b)            # 1/6
print("divide   :", a / b)            # 3/2
print("simplest :", Fraction(6, 8))   # 3/4

print(0.1 + 0.2 == 0.3)                                  # False (float error)
print(Fraction(1, 10) + Fraction(2, 10) == Fraction(3, 10))  # True (exact)
```

#### 3. Ratio and Percentage

**মানে:** `percentage = part ÷ total × 100`। কোনো সংখ্যার percent = `p ÷ 100 × সংখ্যা`।

**Example:** 1200 data-র 80% = `80 ÷ 100 × 1200 = 960` (train)। বাকি 240 (test)। Ratio `960 : 240 = 4 : 1`।

**AI-তে কেন লাগে:** data-কে train আর test-এ ভাগ করতে percentage লাগে।

```python
total = 1200
train = 80 / 100 * total
test = total - train

print("train:", train, "| test:", test)          # 960.0 | 240.0
print("train %:", train / total * 100)           # 80.0
print("ratio train:test =", int(train / test), ": 1")  # 4 : 1
assert train == 960 and test == 240
```

#### 4. Negative Numbers

**মানে:** মাইনাসের মাইনাস হয় প্লাস। `(-) × (-) = +` আর `(+) × (-) = -`।

**Example:** `3 - (-2) = 5`। আর `(-3)^2 = 9`, কিন্তু `-3^2 = -9`।

**AI-তে কেন লাগে:** `w = w - lr × gradient`। gradient negative হলে `w` বাড়ে।

```python
print((-3) ** 2)    # 9
print(-3 ** 2)      # -9
print(-(-5))        # 5
print(3 - (-2))     # 5
print((-2) * (-4))  # 8
assert (-3) ** 2 == 9 and -3 ** 2 == -9 and 3 - (-2) == 5
```

#### 5. Algebra

**মানে:** Variable হলো অজানা সংখ্যা, যেমন `x`। Expression-এ `=` থাকে না, যেমন `3x + 2x` (like terms যোগ করলে `5x`)। Equation-এ `=` থাকে, যেমন `2x + 3 = 11`। দুই পাশে একই কাজ করতে হয়।

**Example:** `2x + 3 = 11` → দুই পাশে 3 বাদ: `2x = 8` → দুই পাশে 2 দিয়ে ভাগ: `x = 4`।

**AI-তে কেন লাগে:** `y = w × x + b` হলো model-এর মূল সূত্র। `w` = weight, `b` = bias।

```python
# 2x + 3 = 11  ->  দুই পাশ থেকে 3 বাদ, তারপর 2 দিয়ে ভাগ
x_solution = (11 - 3) / 2
print("x =", x_solution)            # 4.0
assert 2 * x_solution + 3 == 11

# like terms: 3x + 2x = 5x
x_val = 4
assert 3 * x_val + 2 * x_val == 5 * x_val

# y = weight * x + bias  (দুটি point (2, 5) আর (6, 13) থেকে)
weight = (13 - 5) / (6 - 2)         # slope
bias = 5 - weight * 2               # intercept
print("y =", weight, "* x +", bias)  # y = 2.0 * x + 1.0
```

**আমার code: slope** — `slope = (y2 - y1) ÷ (x2 - x1)`। এটাই `w`।

```python
def slope(x1, y1, x2, y2):
    return (y2-y1)/(x2-x1)

print(slope(2, 5, 6, 13))
print(slope(1, 4, 5, 4))
```

#### 6. Functions

**মানে:** Function হলো একটা মেশিন। input দিলে output দেয়।

**Example:** `f(x) = 2x + 1` হলে `f(3) = 7`। Python-এ: `def f(x): return 2 * x + 1`।

**AI-তে কেন লাগে:** Neural network আসলে একটা বড় function। ছোট function (layer) জোড়া লাগিয়ে বানানো।

```python
def f(x):
    return 2 * x + 1

print(f(3))          # 7

# ছোট "network": layer1 (linear) -> layer2 (ReLU)
def layer1(x):
    return 2 * x + 1

def layer2(x):
    return np.maximum(0, x)   # ReLU: negative হলে 0

def network(x):
    return layer2(layer1(x))

print(network(np.array([-2, 0, 3])))   # [0 1 7]
assert f(3) == 7
```

**আমার code: বাড়ির দাম** — `price = w × area + b`। `w` আর `b` হলো parameter।

```python
# বাড়ির দাম = w * আয়তন + b
w = 6000   # parameter
b = 200000 # parameter

def predict_price(area):
    return w * area + b

print(predict_price(1000))
```

**আমার code: `g(x) = x^2 - 3x`** — `g(-1) = 1 + 3 = 4`। negative `x` বসালে `-3x` প্লাস হয়ে যায় (section 11 দেখো)।

```python
def g(x):
    return pow(x, 2) - 3*x

print(g(0))
print(g(4))
print(g(-1))
```

#### 7. Exponent Rules and Scientific Notation

**মানে:** Exponent মানে বারবার গুণ। Rules:

- `a^m × a^n = a^(m+n)` আর `a^m ÷ a^n = a^(m-n)`
- `(a^m)^n = a^(m×n)`
- `a^0 = 1` আর `a^-n = 1 ÷ a^n`

**Scientific notation (count the jumps):** dot কত ঘর সরে, সেটাই power। ছোট সংখ্যা = negative power, বড় সংখ্যা = positive power।

**Example:** `0.00025` → ৪ ঘর ডানে → `2.5e-4`। `6400000` → ৬ ঘর বামে → `6.4e6`।

**AI-তে কেন লাগে:** learning rate `1e-3`, ছোট probability `1e-10` — সব এভাবে লেখা হয়।

```python
print(2 ** 3 * 2 ** 4 == 2 ** 7)    # True
print(10 ** -3)                      # 0.001
print(2.5e-4 == 0.00025)            # True
print(6.4e6 == 6_400_000)           # True
print(f"{0.00025:.1e}", f"{6400000:.1e}")   # 2.5e-04 6.4e+06
```

#### 8. Logarithm

**মানে:** Log হলো exponent-এর উল্টো। 3টি rule:

- `log(a × b) = log(a) + log(b)`
- `log(a ÷ b) = log(a) - log(b)`
- `log(a^n) = n × log(a)`

`ln` মানে `log` base `e` (`e ≈ 2.718`)। তাই `ln(e^5) = 5`।

**Example:** `log2(8) = 3` কারণ `2^3 = 8`। `log10(1000) = 3`।

**AI-তে কেন লাগে:** loss আর probability-তে log লাগে। log গুণকে যোগ বানিয়ে দেয়।

```python
print(math.log2(8))         # 3.0
print(math.log10(1000))     # 3.0
print(math.log(math.e ** 3))  # 3.0  (ln)

a, b = 8, 4
assert math.isclose(math.log2(a * b), math.log2(a) + math.log2(b))
assert math.isclose(math.log2(a / b), math.log2(a) - math.log2(b))
assert math.isclose(math.log2(a ** 3), 3 * math.log2(a))
```

#### 9. Log Probability in Python

**মানে:** অনেক probability গুণ করলে সংখ্যা খুব ছোট হয়। শেষে `0.0` হয়ে যায় (underflow)। Log নিয়ে যোগ করলে এই সমস্যা হয় না: `log(p1 × p2 × …) = log(p1) + log(p2) + …`

**Example:** `0.1` কে 400 বার গুণ করলে `0.0` আসে। কিন্তু log-এর যোগফল `≈ -921.03`।

**AI-তে কেন লাগে:** পুরো বাক্যের probability অনেক token probability-র গুণ। তাই log-এর যোগ ব্যবহার হয়।

**Experiment 1: ৪টি probability** — সঠিক গুণফল `1e-06`, কিন্তু float একটু ভুল দেয়।

```python
prob_4 = [0.1, 0.05, 0.02, 0.01]

direct_4 = 1
for p in prob_4:
    direct_4 = direct_4 * p
log_sum_4 = sum(math.log(p) for p in prob_4)

print("direct       :", direct_4)               # 1.0000000000000002e-06  (সঠিক: 1e-06)
print("log sum      :", log_sum_4)              # -13.8155...
print("exp(log sum) :", math.exp(log_sum_4))
```

**Experiment 2: ১৪টি সংখ্যা** — ভুল আরও জমে। (১৪টি সংখ্যা আমি নিজে বেছেছি, তোমার notebook-এ ছিল না।)

```python
prob_14 = [0.1, 0.05, 0.02, 0.01, 0.2, 0.3, 0.15,
           0.25, 0.05, 0.08, 0.12, 0.09, 0.07, 0.04]
assert len(prob_14) == 14

direct_14 = 1
exact_14 = Fraction(1)
for p in prob_14:
    direct_14 = direct_14 * p
    exact_14 = exact_14 * Fraction(str(p))      # দশমিক লেখা হুবহু ধরে

print("direct :", direct_14)           # 2.7216000000000007e-16
print("exact  :", float(exact_14))     # 2.7216e-16
print("same?  :", direct_14 == float(exact_14))   # False -> floating-point error
print("log sum:", sum(math.log(p) for p in prob_14))
```

#### 10. Formula Cheat Sheet

| বিষয় | Rule |
|---|---|
| Order | Bracket → Power → Multiply/Divide → Add/Subtract |
| Fraction যোগ | `a/b + c/d = (a×d + c×b) / (b×d)` |
| Fraction গুণ | `a/b × c/d = (a×c) / (b×d)` |
| Fraction ভাগ | `a/b ÷ c/d = a/b × d/c` |
| Percentage | `part ÷ total × 100` |
| Percent of number | `p ÷ 100 × number` |
| Negative | `-(-a) = a`, `(-3)^2 = 9`, `-3^2 = -9` |
| Equation | দুই পাশে একই কাজ করো |
| Linear function | `y = w × x + b` |
| Exponent | `a^m × a^n = a^(m+n)`, `a^m ÷ a^n = a^(m-n)`, `(a^m)^n = a^(m×n)` |
| Zero / negative power | `a^0 = 1`, `a^-n = 1 ÷ a^n` |
| Scientific notation | ছোট → negative power, বড় → positive power |
| Log মানে | `log_b(x) = y` মানে `b^y = x` |
| Log rules | `log(a×b) = log(a)+log(b)`, `log(a÷b) = log(a)-log(b)`, `log(a^n) = n×log(a)` |
| ln | `ln(e^k) = k` |
| Log probability | `log(p1×p2×…) = log(p1)+log(p2)+…` |

```python
x, y, n = 8.0, 4.0, 3
checks = {
    "log(xy) = log x + log y":  np.isclose(np.log(x * y), np.log(x) + np.log(y)),
    "log(x/y) = log x - log y": np.isclose(np.log(x / y), np.log(x) - np.log(y)),
    "log(x^n) = n log x":       np.isclose(np.log(x ** n), n * np.log(x)),
    "a^m * a^n = a^(m+n)":      2 ** 3 * 2 ** 4 == 2 ** 7,
    "a^-n = 1 / a^n":           np.isclose(10.0 ** -3, 1 / 10.0 ** 3),
    "ln(e^5) = 5":              np.isclose(np.log(np.e ** 5), 5),
}
for name, ok in checks.items():
    print("OK " if ok else "BAD", name)
assert all(checks.values())
```

#### 11. Mistakes I made and what to remember

| আমার ভুল | ঠিক | মনে রাখো |
|---|---|---|
| `5 - (-3) = 2` | `5 + 3 = 8` | মাইনাসের মাইনাস = প্লাস |
| `3x + 2x`-এর উত্তর খুঁজেছি | `=` নেই, তাই শুধু `5x` | `=` থাকলে তবেই solve |

```python
# 1. minus of minus
assert 5 - (-3) == 8

# 2. expression vs equation
x = 2
assert 3 * x + 2 * x == 5 * x       # expression: simplify
assert 3 * x + 2 * x == 10          # equation: x = 2 হলে সত্য

# 3. g(x) = x^2 - 3x এ sign
def g_check(x):
    return x ** 2 - 3 * x
print("g(-1) =", g_check(-1))       # 4
assert g_check(-1) == 4

# 4. 15 বনাম 15%
percent = 15 / 100
print("15% of 200 =", percent * 200)   # 30.0
assert math.isclose(percent * 200, 30)
```
