### 1. Order of Operations

**মানে:** আগে Bracket, তারপর Power, তারপর Multiply/Divide, শেষে Add/Subtract।

**Example:** `2 + 3 × 4 = 14` (আগে গুণ)। কিন্তু `(2 + 3) × 4 = 20` (আগে bracket)।

**AI-তে কেন লাগে:** `w * x + b` ভুল ক্রমে করলে model ভুল উত্তর দেয়।

```python
print(2 + 3 * 4)       # 14  (আগে গুণ)
print((2 + 3) * 4)     # 20  (আগে bracket)
print(2 + 3 ** 2)      # 11  (আগে power)
assert 2 + 3 * 4 == 14 and (2 + 3) * 4 == 20 and 2 + 3 ** 2 == 11
```

<a id="sec2"></a>
### 2. Fractions

**মানে:** Fraction হলো ভাগের সংখ্যা। উপরেরটা numerator, নিচেরটা denominator। যোগে denominator সমান করতে হয়, ভাগে উল্টে গুণ করতে হয়।

**Example:** `1/2 + 1/3 = 5/6`, `1/2 × 1/3 = 1/6`, `1/2 ÷ 1/3 = 3/2`, `6/8 = 3/4` (simplest form)।

**AI-তে কেন লাগে:** decimal-এ ছোট ভুল থাকে (`0.1 + 0.2 ≠ 0.3`), exact fraction-এ থাকে না। ছোট ভুল জমে বড় হয়।

```python
a, b = Fraction(1, 2), Fraction(1, 3)
print("add      :", a + b)            # 5/6
print("multiply :", a * b)            # 1/6
print("divide   :", a / b)            # 3/2
print("simplest :", Fraction(6, 8))   # 3/4

print(0.1 + 0.2 == 0.3)                                  # False (float error)
print(Fraction(1, 10) + Fraction(2, 10) == Fraction(3, 10))  # True (exact)
```

### 3. Ratio and Percentage

**মানে:** `percentage = part ÷ total × 100`। কোনো সংখ্যার percent = `p ÷ 100 × সংখ্যা`।

**Example:** 1200 data-র 80% = `80 ÷ 100 × 1200 = 960` (train)। বাকি 240 (test)। Ratio `960 : 240 = 4 : 1`।

**AI-তে কেন লাগে:** data-কে train আর test-এ ভাগ করতে percentage লাগে।

```python
total = 1200
train = 80 / 100 * total
test = total - train

print("train:", train, "| test:", test)          # 960.0 | 240.0
print("train %:", train / total * 100)           # 80.0
print("ratio train:test =", int(train / test), ": 1")  # 4 : 1
assert train == 960 and test == 240
```

### 4. Negative Numbers

**মানে:** মাইনাসের মাইনাস হয় প্লাস। `(-) × (-) = +` আর `(+) × (-) = -`।

**Example:** `3 - (-2) = 5`। আর `(-3)^2 = 9`, কিন্তু `-3^2 = -9`।

**AI-তে কেন লাগে:** `w = w - lr × gradient`। gradient negative হলে `w` বাড়ে।

```python
print((-3) ** 2)    # 9
print(-3 ** 2)      # -9
print(-(-5))        # 5
print(3 - (-2))     # 5
print((-2) * (-4))  # 8
assert (-3) ** 2 == 9 and -3 ** 2 == -9 and 3 - (-2) == 5
```

### 5. Algebra

**মানে:** Variable হলো অজানা সংখ্যা, যেমন `x`। Expression-এ `=` থাকে না, যেমন `3x + 2x` (like terms যোগ করলে `5x`)। Equation-এ `=` থাকে, যেমন `2x + 3 = 11`। দুই পাশে একই কাজ করতে হয়।

**Example:** `2x + 3 = 11` → দুই পাশে 3 বাদ: `2x = 8` → দুই পাশে 2 দিয়ে ভাগ: `x = 4`।

**AI-তে কেন লাগে:** `y = w × x + b` হলো model-এর মূল সূত্র। `w` = weight, `b` = bias।

```python
# 2x + 3 = 11  ->  দুই পাশ থেকে 3 বাদ, তারপর 2 দিয়ে ভাগ
x_solution = (11 - 3) / 2
print("x =", x_solution)            # 4.0
assert 2 * x_solution + 3 == 11

# like terms: 3x + 2x = 5x
x_val = 4
assert 3 * x_val + 2 * x_val == 5 * x_val

# y = weight * x + bias  (দুটি point (2, 5) আর (6, 13) থেকে)
weight = (13 - 5) / (6 - 2)         # slope
bias = 5 - weight * 2               # intercept
print("y =", weight, "* x +", bias)  # y = 2.0 * x + 1.0
```

**আমার code: slope** — `slope = (y2 - y1) ÷ (x2 - x1)`। এটাই `w`।

```python
def slope(x1, y1, x2, y2):
    return (y2-y1)/(x2-x1)

print(slope(2, 5, 6, 13))
print(slope(1, 4, 5, 4))
```

### 6. Functions

**মানে:** Function হলো একটা মেশিন। input দিলে output দেয়।

**Example:** `f(x) = 2x + 1` হলে `f(3) = 7`। Python-এ: `def f(x): return 2 * x + 1`।

**AI-তে কেন লাগে:** Neural network আসলে একটা বড় function। ছোট function (layer) জোড়া লাগিয়ে বানানো।

```python
def f(x):
    return 2 * x + 1

print(f(3))          # 7

# ছোট "network": layer1 (linear) -> layer2 (ReLU)
def layer1(x):
    return 2 * x + 1

def layer2(x):
    return np.maximum(0, x)   # ReLU: negative হলে 0

def network(x):
    return layer2(layer1(x))

print(network(np.array([-2, 0, 3])))   # [0 1 7]
assert f(3) == 7
```

**আমার code: বাড়ির দাম** — `price = w × area + b`। `w` আর `b` হলো parameter।

```python
# বাড়ির দাম = w * আয়তন + b
w = 6000   # parameter
b = 200000 # parameter

def predict_price(area):
    return w * area + b

print(predict_price(1000))
```

**আমার code: `g(x) = x^2 - 3x`** — `g(-1) = 1 + 3 = 4`। negative `x` বসালে `-3x` প্লাস হয়ে যায় (section 11 দেখো)।

```python
def g(x):
    return pow(x, 2) - 3*x

print(g(0))
print(g(4))
print(g(-1))
```

### 7. Exponent Rules and Scientific Notation

**মানে:** Exponent মানে বারবার গুণ। Rules:

- `a^m × a^n = a^(m+n)` আর `a^m ÷ a^n = a^(m-n)`
- `(a^m)^n = a^(m×n)`
- `a^0 = 1` আর `a^-n = 1 ÷ a^n`

**Scientific notation (count the jumps):** dot কত ঘর সরে, সেটাই power। ছোট সংখ্যা = negative power, বড় সংখ্যা = positive power।

**Example:** `0.00025` → ৪ ঘর ডানে → `2.5e-4`। `6400000` → ৬ ঘর বামে → `6.4e6`।

**AI-তে কেন লাগে:** learning rate `1e-3`, ছোট probability `1e-10` — সব এভাবে লেখা হয়।

```python
print(2 ** 3 * 2 ** 4 == 2 ** 7)    # True
print(10 ** -3)                      # 0.001
print(2.5e-4 == 0.00025)            # True
print(6.4e6 == 6_400_000)           # True
print(f"{0.00025:.1e}", f"{6400000:.1e}")   # 2.5e-04 6.4e+06
```

### 8. Logarithm

**মানে:** Log হলো exponent-এর উল্টো। 3টি rule:

- `log(a × b) = log(a) + log(b)`
- `log(a ÷ b) = log(a) - log(b)`
- `log(a^n) = n × log(a)`

`ln` মানে `log` base `e` (`e ≈ 2.718`)। তাই `ln(e^5) = 5`।

**Example:** `log2(8) = 3` কারণ `2^3 = 8`। `log10(1000) = 3`।

**AI-তে কেন লাগে:** loss আর probability-তে log লাগে। log গুণকে যোগ বানিয়ে দেয়।

```python
print(math.log2(8))         # 3.0
print(math.log10(1000))     # 3.0
print(math.log(math.e ** 3))  # 3.0  (ln)

a, b = 8, 4
assert math.isclose(math.log2(a * b), math.log2(a) + math.log2(b))
assert math.isclose(math.log2(a / b), math.log2(a) - math.log2(b))
assert math.isclose(math.log2(a ** 3), 3 * math.log2(a))
```

### 9. Log Probability in Python

**মানে:** অনেক probability গুণ করলে সংখ্যা খুব ছোট হয়। শেষে `0.0` হয়ে যায় (underflow)। Log নিয়ে যোগ করলে এই সমস্যা হয় না: `log(p1 × p2 × …) = log(p1) + log(p2) + …`

**Example:** `0.1` কে 400 বার গুণ করলে `0.0` আসে। কিন্তু log-এর যোগফল `≈ -921.03`।

**AI-তে কেন লাগে:** পুরো বাক্যের probability অনেক token probability-র গুণ। তাই log-এর যোগ ব্যবহার হয়।

**Experiment 1: ৪টি probability** — সঠিক গুণফল `1e-06`, কিন্তু float একটু ভুল দেয়।

```python
prob_4 = [0.1, 0.05, 0.02, 0.01]

direct_4 = 1
for p in prob_4:
    direct_4 = direct_4 * p
log_sum_4 = sum(math.log(p) for p in prob_4)

print("direct       :", direct_4)               # 1.0000000000000002e-06  (সঠিক: 1e-06)
print("log sum      :", log_sum_4)              # -13.8155...
print("exp(log sum) :", math.exp(log_sum_4))
```

**Experiment 2: ১৪টি সংখ্যা** — ভুল আরও জমে। (১৪টি সংখ্যা আমি নিজে বেছেছি, তোমার notebook-এ ছিল না।)

```python
prob_14 = [0.1, 0.05, 0.02, 0.01, 0.2, 0.3, 0.15,
           0.25, 0.05, 0.08, 0.12, 0.09, 0.07, 0.04]
assert len(prob_14) == 14

direct_14 = 1
exact_14 = Fraction(1)
for p in prob_14:
    direct_14 = direct_14 * p
    exact_14 = exact_14 * Fraction(str(p))      # দশমিক লেখা হুবহু ধরে

print("direct :", direct_14)           # 2.7216000000000007e-16
print("exact  :", float(exact_14))     # 2.7216e-16
print("same?  :", direct_14 == float(exact_14))   # False -> floating-point error
print("log sum:", sum(math.log(p) for p in prob_14))
```

**Experiment 3: `[0.1] * 400`** — direct গুণে `0.0` (underflow)। log sum `≈ -921.03`।

```python
import math

prob = [0.1] * 400

result = 1
for p in prob:
    result = result * p
print(result)

log_sum = 0
for p in prob:
    log_sum = log_sum + math.log(p)
print(log_sum)
```

```python
print(400*math.log(0.1))
```

```python
import sys

# (আগের cell-এর result/log_sum আবার আলাদা করে যাচাই)
direct_400 = 1.0
for p in [0.1] * 400:
    direct_400 *= p
log_sum_400 = sum(math.log(p) for p in [0.1] * 400)

print("smallest positive float:", 5e-324, "| smallest normal:", sys.float_info.min)
assert direct_400 == 0.0                                        # underflow
assert math.isclose(log_sum_400, 400 * math.log(0.1))           # log sum ঠিকই আছে
assert math.isclose(log_sum_400, -921.03, abs_tol=0.01)
print("underflow হলো, কিন্তু log sum =", round(log_sum_400, 2))
```

### 10. Formula Cheat Sheet

| বিষয় | Rule |
|---|---|
| Order | Bracket → Power → Multiply/Divide → Add/Subtract |
| Fraction যোগ | `a/b + c/d = (a×d + c×b) / (b×d)` |
| Fraction গুণ | `a/b × c/d = (a×c) / (b×d)` |
| Fraction ভাগ | `a/b ÷ c/d = a/b × d/c` |
| Percentage | `part ÷ total × 100` |
| Percent of number | `p ÷ 100 × number` |
| Negative | `-(-a) = a`, `(-3)^2 = 9`, `-3^2 = -9` |
| Equation | দুই পাশে একই কাজ করো |
| Linear function | `y = w × x + b` |
| Exponent | `a^m × a^n = a^(m+n)`, `a^m ÷ a^n = a^(m-n)`, `(a^m)^n = a^(m×n)` |
| Zero / negative power | `a^0 = 1`, `a^-n = 1 ÷ a^n` |
| Scientific notation | ছোট → negative power, বড় → positive power |
| Log মানে | `log_b(x) = y` মানে `b^y = x` |
| Log rules | `log(a×b) = log(a)+log(b)`, `log(a÷b) = log(a)-log(b)`, `log(a^n) = n×log(a)` |
| ln | `ln(e^k) = k` |
| Log probability | `log(p1×p2×…) = log(p1)+log(p2)+…` |

```python
x, y, n = 8.0, 4.0, 3
checks = {
    "log(xy) = log x + log y":  np.isclose(np.log(x * y), np.log(x) + np.log(y)),
    "log(x/y) = log x - log y": np.isclose(np.log(x / y), np.log(x) - np.log(y)),
    "log(x^n) = n log x":       np.isclose(np.log(x ** n), n * np.log(x)),
    "a^m * a^n = a^(m+n)":      2 ** 3 * 2 ** 4 == 2 ** 7,
    "a^-n = 1 / a^n":           np.isclose(10.0 ** -3, 1 / 10.0 ** 3),
    "ln(e^5) = 5":              np.isclose(np.log(np.e ** 5), 5),
}
for name, ok in checks.items():
    print("OK " if ok else "BAD", name)
assert all(checks.values())
```

### 11. Mistakes I made and what to remember

| আমার ভুল | ঠিক | মনে রাখো |
|---|---|---|
| `5 - (-3) = 2` | `5 + 3 = 8` | মাইনাসের মাইনাস = প্লাস |
| `3x + 2x`-এর উত্তর খুঁজেছি | `=` নেই, তাই শুধু `5x` | `=` থাকলে তবেই solve |
| `g(-1) = 1 - 3 = -2` | `1 + 3 = 4` | negative বসালে bracket দাও |
| `15` লিখেছি | `15% = 0.15` | % মানে `÷ 100` |

```python
# 1. minus of minus
assert 5 - (-3) == 8

# 2. expression vs equation
x = 2
assert 3 * x + 2 * x == 5 * x       # expression: simplify
assert 3 * x + 2 * x == 10          # equation: x = 2 হলে সত্য

# 3. g(x) = x^2 - 3x এ sign
def g_check(x):
    return x ** 2 - 3 * x
print("g(-1) =", g_check(-1))       # 4
assert g_check(-1) == 4

# 4. 15 বনাম 15%
percent = 15 / 100
print("15% of 200 =", percent * 200)   # 30.0
assert math.isclose(percent * 200, 30)
```
