```python
import numpy as np

a = np.array([0.8, 0.7, 0.1])
b = np.array([0.7, 0.8, 0.1])
c = np.array([0.1, 0.0, 0.9])

print(np.dot(a, b))  # Dot product of a and b
print(np.dot(a, c))  # Dot product of a and c
```

```python
A = np.array(
    [[1, 2, 3],
     [4, 5, 6]]
)

print(np.shape(A))  # Shape of matrix A
```

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])
C = np.array([[2, 0],
              [1, 3]])
D = np.array([[1, 4],
              [2, 1]])

print(A + B)  # Element-wise addition of A and B
print(A - B)  # Element-wise subtraction of A and B
print(3 * A)  # Scalar multiplication of A

print(2*C + D)
```

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

M = np.array(
    [[1, 0], [2, 5], [3, 4]]
)

print(A.shape)
print(A.T)
print(A.T.shape)

print(M.shape)
print(M.T)
print(M.T.shape)
    
```

```python
P = np.array([[1, 2], [0, 3]])

Q = np.array([[4, 1], [2, 5]])

print(np.dot(P, Q))
print(P @ Q)  # Using the @ operator for matrix multiplication
```

```python
W = np.array([[1, 2],
              [3, 4],
              [5, 6]])   # shape (3, 2)
x = np.array([10, 1])

print(W.shape)
print(x.shape)
print(W @ x)
```

```python
A = np.array([3,4])
B = np.array([4, 2])

print(A @  B)
print(B @ A)
```
